```python
cols = ['ph', 'Sulfate', 'Trihalomethanes']

print("Mean:")
print(df[cols].mean())

print("\nMedian:")
print(df[cols].median())

print("\nSkewness:")
print(df[cols].skew())
```

```python
for col in cols:
    sns.histplot(data=df, x=col, kde=True)
    plt.title(f'Distribution of {col}')
    plt.show()
```

```python
for col in cols:
    plt.figure(figsize=(6, 3))
    sns.boxplot(x=df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```
